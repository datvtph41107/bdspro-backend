```python
# Import service
import sys
import os

# Lấy thư mục hiện tại (nơi chạy notebook)
current_dir = os.getcwd()
print(f"📁 Current directory: {current_dir}")

# Đảm bảo sử dụng venv Python
venv_path = os.path.join(current_dir, 'venv')
if os.path.exists(venv_path):
    print(f"📦 Found venv at: {venv_path}")
    # Thêm venv site-packages vào path
    # Tìm tất cả các version Python trong venv
    lib_path = os.path.join(venv_path, 'lib')
    if os.path.exists(lib_path):
        for item in os.listdir(lib_path):
            if item.startswith('python'):
                site_packages = os.path.join(lib_path, item, 'site-packages')
                if os.path.exists(site_packages):
                    sys.path.insert(0, site_packages)
                    print(f"✅ Added to path: {site_packages}")
                    break
else:
    print(f"⚠️  venv not found at: {venv_path}")

# Thêm current directory vào path để import app module
sys.path.insert(0, current_dir)

# Kiểm tra pyzbar
try:
    import pyzbar
    print(f"✅ pyzbar available: {pyzbar.__file__}")
except ImportError as e:
    print(f"❌ pyzbar not available: {e}")
    print("💡 Hãy chạy trong terminal:")
    print("   cd ai-service")
    print("   source venv/bin/activate")
    print("   pip install pyzbar")

# Import service
try:
    from app.ocr.easyocr_service import easyocr_service
    print("✅ EasyOCRService imported successfully")
except ImportError as e:
    print(f"❌ Failed to import EasyOCRService: {e}")
    print(f"   Python path: {sys.executable}")
    print(f"   sys.path: {sys.path[:3]}...")
```

```python
# Test đọc QR code từ URL
# Thay URL bằng link ảnh QR code thực tế
image_url = ""  # Thay bằng URL thực tế

print(f"📥 Downloading image from: {image_url}")
results = easyocr_service.read_qr_code_from_url(image_url)

if results:
    print(f"\n✅ Found {len(results)} QR code(s):\n")
    for i, result in enumerate(results, 1):
        print(f"QR Code #{i}:")
        print(f"  - Data: {result['data']}")
        print(f"  - Type: {result['type']}")
        print(f"  - Position: x={result['rect']['x']}, y={result['rect']['y']}, width={result['rect']['width']}, height={result['rect']['height']}")
        print()
else:
    print("❌ No QR code found in the image")
```
