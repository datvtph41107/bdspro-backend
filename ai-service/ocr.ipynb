```python
# Import MRZ Scan Service
import sys
import os

# Add project root to path
sys.path.insert(0, os.path.abspath('.'))

from app.mrz.mrz_scan_service import MRZScanService

# Khởi tạo service
mrz_service = MRZScanService()

print("✅ MRZ Scan Service initialized")
```

```python

# Test với URL ảnh CCCD mặt sau
# Thay đổi URL này bằng URL thực tế của bạn
image_url = ""  # Thay bằng URL thực tế

print(f"📸 Reading MRZ from URL: {image_url}")
print("-" * 60)

# Đọc MRZ
result = mrz_service.read_mrz_from_url(image_url)

# In kết quả
if result:
    print("\n✅ MRZ đọc thành công!")
    print(f"\n📋 Method used: {result.get('method', 'unknown')}")
    print(f"\n📄 Raw MRZ:")
    print(result.get('raw', 'N/A'))
    
    print(f"\n🔍 Parsed Information:")
    parsed = result.get('parsed', {})
    if parsed:
        print(f"   Document Type: {parsed.get('document_type', 'N/A')}")
        print(f"   Full Name: {parsed.get('full_name', 'N/A')}")
        print(f"   Country Code: {parsed.get('country_code', 'N/A')}")
        print(f"   ID Number: {parsed.get('id_number', 'N/A')}")
        print(f"   Date of Birth: {parsed.get('date_of_birth', 'N/A')}")
        print(f"   Expiry Date: {parsed.get('expiry_date', 'N/A')}")
    else:
        print("   No parsed information available")
else:
    print("\n❌ Không thể đọc MRZ từ ảnh")
    print("   Có thể do:")
    print("   - URL không hợp lệ hoặc không thể truy cập")
    print("   - Ảnh không rõ ràng")
    print("   - Ảnh không chứa MRZ")
```
