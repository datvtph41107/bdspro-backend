```python
import qrcode
from pyzbar.pyzbar import decode
from PIL import Image
import json
import os

# Danh sách các ký tự tiếng Việt bạn muốn map
viet_chars_list = list("a à á ả ã ạ ă ằ ắ ẳ ẵ ặ â ầ ấ ẩ ẫ ậ đ e è é ẻ ẽ ẹ ê ề ế ể ễ ệ i ì í ỉ ĩ ị o ò ó ỏ õ ọ ô ồ ố ổ ỗ ộ ơ ờ ớ ở ỡ ợ u ù ú ủ ũ ụ ư ừ ứ ử ữ ự y ỳ ý ỷ ỹ ỵ")

# Folder tạm để lưu QR code
os.makedirs("tmp_qr", exist_ok=True)

mapping = {}

for char in viet_chars_list:
    # Tạo QR code cho từng ký tự
    qr = qrcode.QRCode(
        version=2,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=10,
        border=4
    )
    qr.add_data(char)
    qr.make(fit=True)
    img = qr.make_image(fill_color="black", back_color="white")

    # Lưu tạm file QR code
    temp_file = f"tmp_qr/{ord(char)}.png"
    img.save(temp_file)

    # Decode QR code
    img_dec = Image.open(temp_file)
    decoded_objects = decode(img_dec)

    if decoded_objects:
        raw_bytes = decoded_objects[0].data
        # Lưu mapping dạng string escape sequence
        mapping[char] = list(raw_bytes)  # list of ints để JSON hóa dễ dàng
    else:
        print(f"Không decode được ký tự: {char}")

# Lưu vào file JSON
with open("viet_bytes_map.json", "w", encoding="utf-8") as f:
    json.dump(mapping, f, ensure_ascii=False, indent=2)

print("Mapping lưu xong vào viet_bytes_map.json")
```

```python
import json
from pyzbar.pyzbar import decode
from PIL import Image

# Load file JSON mapping
with open("viet_bytes_map.json", "r", encoding="utf-8") as f:
    mapping = json.load(f)

# Chuyển list of ints thành dict bytes -> char
bytes_to_char = {}
for char, byte_list in mapping.items():
    bytes_to_char[bytes(byte_list)] = char

def decode_qr_with_map(img_path):
    img = Image.open(img_path)
    decoded_objects = decode(img)

    results = []
    for obj in decoded_objects:
        raw_bytes = obj.data
        # Nếu raw_bytes đúng với mapping, lấy char từ map
        # Hoặc thử từng ký tự trong raw_bytes
        print(raw_bytes)
        decoded_text = ""
        i = 0
        while i < len(raw_bytes):
            matched = False
            # Duyệt tất cả key trong map để check prefix
            for b_seq, char in bytes_to_char.items():
                seq_len = len(b_seq)
                if raw_bytes[i:i+seq_len] == b_seq:
                    decoded_text += char
                    i += seq_len
                    matched = True
                    break
            if not matched:
                # Nếu không tìm thấy trong map, decode bình thường
                try:
                    decoded_text += raw_bytes[i:i+1].decode("utf-8")
                except UnicodeDecodeError:
                    decoded_text += "?"  # ký tự lạ
                i += 1
        results.append(decoded_text)
    return results

# Ví dụ test
decoded = decode_qr_with_map("viet_chars.png")
print("Decoded with map:", decoded)
```
